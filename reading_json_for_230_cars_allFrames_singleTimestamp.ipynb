```python
def get_data():
    
    question_access = read_content["Sequence"]
    
    for question_data in question_access:
        replies_access = question_data['Labels']
        
        for replies_data in replies_access:
            user_name = replies_data['Devices']
            
            for Channels_data in user_name:
                channels_name = Channels_data['Channels']
                
                for Object_data in channels_name:
                    object_names = Object_data['ObjectLabels']
                    
                    for user in object_names: 
                        userdata0 = user["TimeStamp"]
                        a=('MFC4xxLongImageRight'+ str(userdata0)+ '.jpeg')
                        userdata1 = user['FrameNumber']
                        userdata2 = user['FrameObjectLabels']                        
                        
                        n=len(userdata2)                        
                        for i in range(n):
                            
                            print("TimeStamp:",a)
                            print("FrameNumber:",userdata1)
                            
                            userdata2_1 = user['FrameObjectLabels'][i]['category']
                            print("category:",userdata2_1)

                            userdata2_2 = user['FrameObjectLabels'][i]['Trackid']
                            print("Trackid:",userdata2_2)

                            userdata2_3 = user['FrameObjectLabels'][i]['width']
                            print("width:",userdata2_3)

                            userdata2_4 = user['FrameObjectLabels'][i]['height']
                            print("height:",userdata2_4)

                            userdata2_5 = user['FrameObjectLabels'][i]['shape']['x'][0]
                            print("x:",userdata2_5)

                            userdata2_6 = user['FrameObjectLabels'][i]['shape']['y'][0]
                            print("y:",userdata2_6)
                            
                            TimeStamp.append(a)
                            FrameNumber.append(userdata1)
                            category.append(userdata2_1)
                            Trackid.append(userdata2_2)
                            width.append(userdata2_3)
                            height.append(userdata2_4)
                            x.append(userdata2_5)
                            y.append(userdata2_6)            
```

```python
TimeStamp, FrameNumber, category, Trackid, width, height, x, y = [],[],[],[],[],[],[],[]
```

```python
b=get_data()
print(b)
```

```python
import glob, os, json

json_dir = 'C:\\Users\\uif68352\\Active_Learning\\Labelled_230_car_images\\Jsons_and_csv'

json_pattern = os.path.join(json_dir, '*.json')
file_list = glob.glob(json_pattern)
print(file_list)
```

```python
import pandas as pd
dfs = []
for file in file_list:
    df_temp = pd.read_json(file,  lines=True)
    df_temp['source'] = file.rsplit("/", 1)[-1]
    dfs.append(df_temp)
df = pd.concat(dfs)
df = df.T
```

```python
df.columns=['TimeStamp','FrameNumber','category','Trackid','width','height','x','y']
df = df.iloc[:-1]
display(df)
```
