```python
import os
import numpy as np
from PIL import Image

folder_path = "runs/detect/exp3/"  # Replace with the path to your folder
images_per_row = 5  # Change this to the desired number of images per row

# Get a list of all the files in the folder
file_list = os.listdir(folder_path)

# Create an empty list to store the images
image_list = []

# Loop through each file and append the image to the list
for filename in file_list:
    if filename.endswith(".jpeg") or filename.endswith(".png"):  # Only append image files
        file_path = os.path.join(folder_path, filename)
        with Image.open(file_path) as img:
            image_list.append(np.array(img))

# Determine the number of rows and columns in the final image
num_images = len(image_list)
num_rows = num_images // images_per_row + int(num_images % images_per_row != 0)
num_cols = min(images_per_row, num_images)

# Create a blank image to hold the final result
result_width = image_list[0].shape[1] * num_cols
result_height = image_list[0].shape[0] * num_rows
result = np.zeros((result_height, result_width, 3), dtype=np.uint8)

# Loop through each image and paste it into the result image
for i, img in enumerate(image_list):
    row = i // images_per_row
    col = i % images_per_row
    x = col * img.shape[1]
    y = row * img.shape[0]
    result[y:y+img.shape[0], x:x+img.shape[1], :] = img

# Convert the result back to a PIL image object
result = Image.fromarray(result)

# Display the result
result.show()
```
